# 2. Statistical Imputation (Mean/Median/Mode)

```python
from sklearn.impute import SimpleImputer
```

```python
num_imputer=SimpleImputer(strategy='median')
df['age']=num_imputer.fit_transform(df[['age']])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df["embarked"].value_counts()
```

```python
cat_imputer=SimpleImputer(strategy='most_frequent')
df['embarked']=cat_imputer.fit_transform(df[['embarked']]).ravel()
```

```python
df.isnull().sum()
```

# 3. Multivariate Imputation (Iterative Imputer)

```python
df.head()
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

it_imputer = IterativeImputer(random_state=42)
# We use relevant numerical columns to estimate the age
df[['age', 'pclass', 'fare']] = it_imputer.fit_transform(df[['age', 'pclass', 'fare']])
```

```python
df.isnull().sum()
```

# 4. KNN Imputation

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import KNNImputer
```

```python
knn_imputer = KNNImputer(n_neighbors=5)
df["age"] = knn_imputer.fit_transform(df[["age"]])
```
