```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer  #Missing value handle 
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv("covid_toy.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],
                                                test_size=0.2, random_state=43)
```

```python
x_train.head()
```

```python
x_train.isnull().sum()
```

```python
# Adding simple imputer to fever col
si=SimpleImputer()
x_train_fever=si.fit_transform(x_train[['fever']])

# also the dataset 
x_test_fever=si.transform(x_test[['fever']])
x_train_fever.shape,x_test_fever.shape
```

```python
#x_train_fever
```

```python
# Ordinalencoding -> cough
ore=OrdinalEncoder(categories=[['Mild','Strong']])
x_train_cough=ore.fit_transform(x_train[['cough']])

x_test_cough=ore.transform(x_test[['cough']])

x_train_cough.shape, x_test_cough.shape
```

```python
x_train_cough
```

```python
## OneHotEncoding -> gender, city
ohn=OneHotEncoder(drop='first',sparse_output=False)
x_train_city=ohn.fit_transform(x_train[['gender','city']])

# also the test data 
x_test_city=ohn.transform(x_test[['gender','city']])

x_train_city.shape, x_test_city.shape
```

### column Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False, drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
x_train.head()
```

```python
new_train_data=transformer.fit_transform(x_train)
```

```python
new_train_data
```
